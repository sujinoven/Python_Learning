# 1.Function example for all Python Data Structures

## 1.1. LIST

### Append - Adds an element at the end.

```python
items = [10,20,34,32,45]

items.append(10)

items
```

### Clear - Removes all elements.


```python
items.clear()
items
```

### Copy - Creates a copy of the list.


```python
items_1 = [10,20,34,32,45]

items_2 =items_1.copy()


items_2
```

### extend() - Adds multiple elements to the list.

```python
a = [1, 2]
a.extend([3, 4])
print(a)
```

### insert() - Inserts element at a specific position.

```python
a = [1, 2, 3]
a.insert(1, 10)
print(a)  
```

### remove() - Removes the first matching element.

```python
a = [1, 2, 3, 2]
a.remove(2)
print(a) 
```

### pop() - Removes and returns element by index.

```python
a = [1, 2, 3]
a.pop(1)
print(a)
```

### count() - Counts occurrences of an element.

```python
a = [1, 2, 2, 3]
print(a.count(2))
```

### sort() - Sorts the list.

```python
a = [3, 1, 2]
a.sort()
print(a) 
```

### reverse() -  Reverses the list.

```python
a = [1, 2, 3]
a.reverse()
print(a) 
```

## 2.TUPLE

## count()- Counts how many times an element appears.

```python
a = (1, 2,1, 3)
print(a.count(1))
```

## index() - Returns the position (index) of the first occurrence.

```python
t = (10, 20, 30)
print(t.index(20))
```

## 3.SET

### add() - Adds one element to the set.

```python
s = {1, 2}
s.add(3)
print(s)
```

### update() -  Adds multiple elements (from another set/list/tuple).

```python
s = {1, 2}
s.update([3, 4])
print(s) 
```

### remove()-Removes an element (error if not found).

```python
s = {1, 2, 3}
s.remove(2)
print(s)
```

### discard()- Removes element (no error if not found).

```python
s = {1, 2, 3}
s.discard(4)
```

### pop()- Removes a random element.

```python
s = {1, 2, 3}
s.pop()
print(s)
```

### clear() - Removes all elements.

```python
s = {1, 2, 3}
s.clear()
print(s)
```

### union() - Combines two sets.

```python
a = {1, 2}
b = {3, 4}
print(a.union(b))
```

### intersection() - Returns common elements.

```python
a = {1, 2, 3}
b = {2, 3, 4}
print(a.intersection(b))
```

### difference() - Returns elements in first set but not in second.

```python
a = {1, 2, 3}
b = {2, 4}
print(a.difference(b))
```

### symmetric_difference() - Returns elements not common in both.

```python
a = {1, 2, 3}
b = {2, 3, 4}
print(a.symmetric_difference(b))
```

### issubset() - Checks if one set is a subset of another.

```python
a = {1, 2}
b = {1, 2, 3}
print(a.issubset(b)) 
```

### issuperset() - Checks if one set contains another.

```python
a = {1, 2, 3}
b = {1, 2}
print(a.issuperset(b))
```

### isdisjoint() - Checks if sets have no common elements.

```python
a = {1, 2}
b = {3, 4}
print(a.isdisjoint(b))
```

### copy() - Creates a copy of the set.

```python
s = {1, 2, 3}
s2 = s.copy()
print(s2)
```

## DICTIONARY

### get() - Returns value for a key (no error if key missing).

```python
d = {'a': 1, 'b': 2}
print(d.get('a')) 
```

### keys() - Returns all keys.

```python
d = {'a': 1, 'b': 2}
print(d.keys()) 
```

### values() - Returns all values.

```python
d = {'a': 1, 'b': 2}
print(d.values())
```

### items() - Returns key-value pairs.

```python
d = {'a': 1, 'b': 2}
print(d.items())  
```

### setdefault() - Returns value; if key not present, inserts it.

```python
d = {'a': 1}
d.setdefault('b', 2)
print(d)
```

### update() - Updates dictionary with new key-value pairs.

```python
d = {'a': 1}
d.update({'b': 2})
print(d)
```

### pop() - Removes a key and returns its value.

```python
d = {'a': 1, 'b': 2}
d.pop('a')
print(d)
```

### popitem() - Removes last inserted item.

```python
d = {'a': 1, 'b': 2}
d.popitem()
print(d)
```

### clear() - Removes all elements.

```python
d = {'a': 1}
d.clear()
print(d)  
```

### copy() - Creates a copy of the dictionary.

```python
d = {'a': 1, 'b': 2}
d2 = d.copy()
print(d2)
```

### fromkeys() - Creates a dictionary from given keys.

```python
keys = ['a', 'b', 'c']
d = dict.fromkeys(keys, 0)
print(d)
```
