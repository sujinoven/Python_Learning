# Creating Data Frames using Dictionaries

### 1.Student DataFrame

```python
import pandas as pd

student_data={
    "Name":["Sujin","Divo","Ani","Jerald","Rani"],
    "Maths":[96,97,98,99,100],
    "Science":[99,98,97,96,98]
            }
df = pd.DataFrame(student_data)

df
```

## 2.Employee DataFrame

```python
import pandas as pd

employee_data={
    "Name":["Sujin","Divo","Ani","Jerald","Rani"],
    "Department":["IT","HR","Finance","HR","IT"],
    "Salary":[10000,20000,30000,40000,45000]
}


df = pd.DataFrame(employee_data)
df
```

### 3.Product Sales Data

```python
product_sales_data = {
    "Product":["Laptop","TV","Mobile","Tablet"],
    "price":[70000,40000,20000,30000],
    "Quantity Sold":[10,8,25,16]
}

df = pd.DataFrame(product_sales_data)

df
```
