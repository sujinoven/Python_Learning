### 1. Which Category Generates the Highest Sales?

```python
superstore.head()
```

```python
## 1. Which Category Generates the Highest Sales?

category_sales = superstore.groupby("Category")["Sales"].sum().sort_values(ascending=False).round(2)

category_sales
```

```python
##Visualization
plt.figure(figsize=(5,5))

sns.barplot(x=category_sales.index,y=category_sales.values)


plt.title('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```

### We could see that the Technology Category generated the highest sales followed with Furniture Category and Office Supplies

## 2. Which Category Generates the Highest Profit?


```python
category_profit = superstore.groupby("Category")["Profit"].sum().sort_values(ascending=False).round(2)
category_profit
```

```python
#Visualization

plt.figure(figsize=(5,5))

sns.barplot(x=category_profit.index,y=category_profit.values)


plt.title('Profit by Category')
plt.xlabel('Category')
plt.ylabel('Profit')
plt.show()
```

### The Technology category generated the highest profit, followed by Furniture, while Office Supplies contributed the least. This trend highlights that higher sales within a category are directly associated with higher profit..

## 3. Which Region Has the Highest Sales?

```python
regions_sales = superstore.groupby("Region")["Sales"].sum().sort_values(ascending=False).round(2)
regions_sales
```

```python
plt.figure(figsize=(5,5))

sns.barplot(x=regions_sales.index,y=regions_sales.values)

plt.title("Region Wise Sales")
plt.xlabel("Regions")
plt.ylabel("Sales")
plt.show()
```

### The West region achieved the highest sales, followed by the East, while the Central and South regions recorded comparatively lower figures. This pattern reinforces the observation that higher sales in a region are directly linked to higher profit generation.

## 4. Top 10 States by Sales

```python
top_states = superstore.groupby('State')['Sales'].sum() \
               .sort_values(ascending=False) \
               .head(10).round(2)

print(top_states)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(y=top_states.index,
            x=top_states.values,
            palette='Blues_r')

plt.title('Top 10 States by Sales')
plt.xlabel('Sales')
plt.ylabel('State')
plt.show()
```

### California recorded the highest sales, followed by New York and Texas, while states such as Virginia, Michigan, and Ohio contributed comparatively lower figures. This distribution underscores the principle that higher sales in a state are directly associated with higher profit.

## 5.Which Customer Segment Generates Maximum Revenue?

```python
segment_sales = superstore.groupby('Segment')['Sales'].sum()

print(segment_sales)
```

```python
plt.figure(figsize=(7,7))

plt.pie(segment_sales,
        labels=segment_sales.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title('Sales Contribution by Segment')
plt.show()
```

### The Consumer segment accounts for the largest share of sales at 50.6%, followed by Corporate with 30.7%, and Home Office with 18.7%. This distribution reinforces the principle that higher sales in a segment are directly associated with higher profit generation.

### 6. Top 10 Products by Sales


```python
top_products = superstore.groupby('Product Name')['Sales'] \
                 .sum() \
                 .sort_values(ascending=False) \
                 .head(10)

print(top_products)
```

```python
plt.figure(figsize=(12,6))

sns.barplot(y=top_products.index,
            x=top_products.values,
            palette='rocket')

plt.title('Top 10 Products by Sales')
plt.xlabel('Sales')
plt.ylabel('Products')
plt.show()
```

###  The Canon imageCLASS 2200 Advanced Copier leads all products in sales, followed by the Fellowes PB500 Electric Punch Plastic Comb Binding Machine and the Cisco TelePresence System EX90 Videoconferencing Unit. Overall, the data reveals that office equipment and technology‑related products dominate sales performance, reinforcing the trend that higher sales volumes correspond to higher profitability.
