## 1.Abstraction
Abstraction means hiding implementation details and exposing only what is necessary.
Suppose every delivery vehicle must have a deliver_order() method, but each vehicle delivers differently.

```python
from abc import ABC,abstractmethod

class Delivery_Vehicle(ABC):

    @abstractmethod
    def deliver_order(self):
        pass

#Now different vehicles provide their own implementation.

class Car(Delivery_Vehicle):

    def deliver_order(self):
        print("Delivering the Order using Car")

class Bike(Delivery_Vehicle):

    def deliver_order(self):
        print("Delivering the Order using Bike")

bike=Bike()
bike.deliver_order()
    
```

### Why is this abstraction?
The food delivery system only knows:
Pythondeliver_order()

It doesn't care how the bike or drone performs the delivery.

## 2.Inheritance
Inheritance means a child class reuses properties and methods from a parent class.
All delivery vehicles have common information:

vehicle number,
speed,
tracking

Instead of writing them repeatedly:

```python
class Delivery_Vehicle:
    def __init__(self,vehicle_number,speed):
        self.vehicle_number=vehicle_number
        self.speed=speed

    def vehicle_tracking(self):
        print(f"Tracking {self.vehicle_number}")

# Now Bike and Car inherit from it.

class Bike(Delivery_Vehicle):
    pass
class Car(Delivery_Vehicle):
    pass

bike1=Bike(1234,45)
bike1.vehicle_tracking()
```

### Why is this inheritance?
Because:
* Bike is a DeliveryVehicle
* Car is a DeliveryVehicle

They automatically get all common functionality.
