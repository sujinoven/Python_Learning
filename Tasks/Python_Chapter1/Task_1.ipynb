# Task 1: 2 Examples of all String functions

## 1.upper() (Convert to uppercase)

```python
text = "hello"
print(text.upper())
```

```python
name = "How are you?"
print(name.upper())
```

## 2.lower() (Convert to lowercase)

```python
text = "hello"
print(text.lower())
```

```python
name = "How are you?"
print(name.lower())
```

## 3.strip() (Remove spaces from ends)

```python
text = "  hello  "
print(text.strip())
```

```python
data = "   Python   "
print(data.strip())
```

## 4. replace() (Replace substring)

```python
text = "I like Java"
print(text.replace("Java", "Python"))
```

```python
msg = "apple apple"
print(msg.replace("apple", "mango"))
```

## 5. split() (Split string into list)


```python
text = "apple,banana,orange"
print(text.split(","))
```

```python
sentence = "Python is easy"
print(sentence.split())
```

## 6. join() (Join list into string)


```python
words = ["Python", "is", "fun"]
print(" ".join(words))
```

```python
chars = ['P', 'y', 't', 'h', 'o', 'n']
print("-".join(chars))
```

## 7. find() (Find position of substring)


```python
text = "hello world"
print(text.find("world"))
```

```python
word = "python programming"
print(word.find("gram"))
```

## 8. count() (Count occurrences)


```python
text = "apple apple apple"
print(text.count("apple"))
```

```python
word = "banana"
print(word.count("a"))
```

## 9.Capitalize

```python
print("hello world".capitalize())
print("python".capitalize())
```

## 10.swapcase()

```python
print("Hello".swapcase())
print("PyThOn".swapcase())
```

## 11.center()

```python
print("hi".center(10, "*"))
print("python".center(12, "-"))
```

## 12.ljust()

```python
print("hi".ljust(10, "-"))
print("cat".ljust(8, "*"))
```

## 13.rjust()

```python
print("hi".rjust(10, "-"))
print("dog".rjust(8, "*"))
```

## 14.strip()

```python
print("  hello  ".strip())
print("***hi***".strip("*"))
```

## 15.lstrip()

```python
print("  hello".lstrip())
print("***hi".lstrip("*"))
```

## 16.rstrip()

```python
print("hello  ".rstrip())
print("hi***".rstrip("*"))
```

## 17.rfind()

```python
print("hello hello".rfind("hello"))
print("banana".rfind("a"))
```

## 18.index()

```python
print("hello".index("e"))
print("python".index("t"))
```

## 19.rindex()

```python
print("hello hello".rindex("hello"))
print("banana".rindex("a"))
```

## 20.isalpha()

```python
print("Hello".isalpha())
print("Hello123".isalpha())
```

## 21.isdigit()

```python
print("123".isdigit())
print("123a".isdigit())
```

## 22.isalnum()

```python
print("Hello123".isalnum())
print("hello!".isalnum())
```

## 23.isspace()


```python
print("   ".isspace())
print(" a ".isspace())
```

## 24.islower()

```python
print("hello".islower())
print("Hello".islower())
```

## 25.isupper()

```python
print("HELLO".isupper())
print("Hello".isupper())
```

## 26.istitle()

```python
print("Hello World".istitle())
print("hello world".istitle())
```

## 27.expandtabs()

```python
print("a\tb".expandtabs(4))
print("1\t2\t3".expandtabs(2))
```

## 28.splitlines()

```python
print("hello\nworld".splitlines())
print("a\nb\nc".splitlines())
```

## 29.count()

```python
print("apple apple".count("apple"))
print("banana".count("a"))
```

## 30.Starts With

```python
print("hello".startswith("he"))
print("python".startswith("py"))
```

## 31.Ends With

```python
print("hello".endswith("lo"))
print("python".endswith("on"))
```

## 32.partition()

```python
print("hello-world".partition("-"))
print("abc@xyz".partition("@"))
```

## 32.rpartition()

```python
print("a-b-c".rpartition("-"))
print("1@2@3".rpartition("@"))
```

## 33.encode()

```python
print("hello".encode())
print("python".encode("utf-8"))
```

## 34. zfill()

```python
print("5".zfill(3))
print("50".zfill(5))
```
