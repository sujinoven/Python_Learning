# 1.While Loop

## 1.1 Write a program to print even numbers from 1 to 20 using a while loop.


```python
num = 2

while num <= 20:
    print(num)
    num+=2
```

## 1.2.Write a program to find the sum of first 10 natural numbers using a while loop.

```python
num =1
sum= 0

while num<=10:
    sum+=num
    num+=1
print("Sum:",sum)
```

# 4.If-Elif else

## 4.1.Write a program to check whether a number is positive, negative, or zero using if–elif–else.


```python
num = int(input("Enter the Number"))

if num > 0:
    print("Positive")
elif num<0:
    print("Negative")
else:
    print("Zero")
```

## 4.2.Write a program to find the largest among three numbers using if–elif–else.


```python
a = int(input("Enter a:"))
b=int(input("Enter b:"))
c=int(input("Enter c:"))

if a>b and a>c:
    print("a is largest")
elif b>a and b>c:
    print("b is largest")
else:
    print("c is largest")
```

# 6.User defined functions with for loop

## 6.1.Write a user-defined function that takes a number as input and prints the multiplication table of that number using a for loop.

```python
def multiplication_table(number):

    for i in range(1,11):
        i*=num
        print(i)

num=int(input("Enter the Number:"))
multiplication_table(num)
```

## 6.2.Write a user-defined function to calculate and return the sum of all numbers from 1 to n using a for loop.


```python
def sum_of_num(n):
    sum =0

    for i in range(1,n+1):
        sum+=i
        print(sum)

num=int(input("Enter the Number:"))
sum_of_num(num)
```

# 7.user defined functions with for loop and if condition

## 7.1.Write a user-defined function that takes a number as input and prints all even numbers from 1 to that number using a for loop and an if condition.

```python
def even_num(num):

    for i in range(1,num+1):
        if i%2==0:
            print(i)
            i+=1
n=int(input("Enter the Number:"))
even_num(n)
```

## 7.2.Write a user-defined function to count how many odd numbers are present between 1 and n using a for loop and an if condition, and return the count.

```python

def count_odd(num):
    count = 0

    for i in range(1,num+1):
        if i%2==0:
            count+=1
            
    print(count)
n=int(input("Enter the Number:"))
count_odd(n)
```
