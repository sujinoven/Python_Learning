### Tope 8 Powerful functions in Python for Deeper Data Analysis

### 1.groupby

```python
import pandas as pd

insurance = pd.read_csv("Insurance.csv")
insurance
```

```python
insurance.isna().sum()
```

```python
insurance.dtypes
```

### Pull Up the average values based on region

```python
insurance.groupby("region")["charges"].mean().round(2).sort_values()
```

### Pull Up the Average values based on Region and Sex

```python
insurance.head()
```

```python
insurance.groupby(["region","sex"])["charges"].mean().round(2).sort_values()
```

### 2.Pivot Table

```python
pd.pivot_table(data=insurance,values="charges",index="region",columns=["sex","children"]).round(2)
```

### 3.Aggregate

Pull Up the Sum,Min.max.mean values based on Region and Sex

```python
insurance.groupby("region")["charges"].agg(["min","max","mean","sum"]).round(2)
```

### 4.Crosstab

How many smokers and non smokers in a region

```python
pd.crosstab(index=insurance["region"],columns=insurance["smoker"],margins=True)
```

### 6.CONCAT

```python
sales_2017 = pd.read_csv("Sales_Transactions_2017.csv")
sales_2018 = pd.read_csv("Sales_Transactions_2018.csv")
sales_2019 = pd.read_csv("Sales_Transactions_2019.csv")
```

```python
sales_concat = pd.concat([sales_2017,sales_2018,sales_2019])
```

```python
sales_concat
```

### 7.Apply

Create a New Column as "Eligibility" and mark as Eligible if age > 20 else Not Eligible

```python
def get_age(age):
    if age > 20:
        return "Eligible"
    else:
        return "Not Eligible"
```

```python
insurance["Eligibility"] = insurance["age"].apply(get_age)
```

```python
insurance
```

### 8.Lambda 

```python
insurance["eligibility_2"] = insurance["age"].apply(lambda x:1 if x>20 else 0)
```

```python
insurance
```
