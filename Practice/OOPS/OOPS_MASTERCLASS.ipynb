## init,del,str

```python
class Bank_Account():

    def __init__(self,customer_name,account_number,balance):
        self.customer_name=customer_name
        self.account_number=account_number
        self.balance=balance

    def __str__(self):
        return self.customer_name

    def deposit(self):
        print("The amount has been credited")
    def debit(self):
        print("The amount has been debited")

account_1=Bank_Account("Sujin",2484894,200000)

print(account_1)
```

### 1.Encapsulation -- Binding of variables and methods inside the class
Getter and Setter

```python
class Bank_Account():
    def __init__(self,name,account_number,balance):
        self.name= name
        self.account_number=account_number
        self.__balance=balance
        
    def deposit(self):
        print("The amount has been credited")
    def debit(self):
        print("The amount has been debited")

    #getter

    def get_balance(self):
        return self.__balance

    ## Setter

    def set_balance(self,balance):
        self.__balance=balance


user_1=Bank_Account("SUJIN",123456,200000)
user_1.name
user_1.set_balance(100000)
user_1.get_balance()
    
```

### 2.Inheritance -- It is a class When One class (Child) inheriting the properties and methods of another class (Parent).

* 1.Singel/ Inheritance     : One Parent -> One Child
* 2.Multi level Inheritance  : Parent -> Child -> Grand Child
* 3.Multiple Inheritance     : Child  Inherits  from 2 Parents
* 4.Hierarchial Inheritance  : One Parent -> Multiple Kids
* 5.Hybrid Inheritance       : Combination of different types

#### 1.Single Inheritance     : One Parent -> One Child

```python
class Vehicle():
    name="Toyota"
    no_of_wheels=6
    no_of_airbags=5

    def move_forward(self):
        return "The vehicle is moving forward"

class Car(Vehicle):

    def __init__(self,mileage):
        self.mileage=mileage

    def moving_speed(self):
        return "The {self.name} is moving with the speed of 20km/hr"

car1=Car(25)
car1.name
```

### 2.Multi level Inheritance  : Parent -> Child -> Grand Child

```python
class A:
    pass
class B(A):
    pass
class C(B):
    pass
```

### 3.Multiple Inheritance     : Child  Inherits  from 2 Parents

```python
class A:
    pass
class B:
    pass
class C(A,B):
    pass

## If botht A and B, has same attributes or parents, it'll return the first class attribute/method which is inside the bracket
```

####  4.Hierarchial Inheritance  : One Parent -> Multiple Kids

```python
class A:
    pass
class B(A):
    pass
class C(A):
    pass
```
