### Univariate Analysis

```python
# Visualizing the Data

import matplotlib.pyplot as plt

plt.hist(mtcars["mpg"])

plt.xlabel("Miles per Gallon")
plt.ylabel("Frequency")
plt.title("MPG DISTRIBUTION")
plt.show()
```

```python
plt.boxplot(mtcars["mpg"])
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.hist(mtcars["disp"])

plt.xlabel("Disp")
plt.ylabel("Frequency")
plt.title("Disp Over Frequency")

plt.show()
```

```python
import numpy as np

students = ["A","B","C"]
math = [80,70,90]
science = [75,85,88]

x = np.arange(len(students))

plt.bar(x-0.2, math, width=0.4, label='Math')
plt.bar(x+0.2, science, width=0.4, label='Science')

plt.xticks(x, students)
plt.legend()
plt.show()
```

```python
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    "Hours":[1,2,3,4,5],
    "Marks":[40,50,60,70,80],
    "Attendance":[70,75,80,90,95]
})

sns.heatmap(df.corr(),
            annot=True,
            cmap="coolwarm")
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

hours = [1,2,3,4,5]
marks = [40,50,60,70,80]
attendance = [70,75,80,85,90]

ax.scatter(hours, marks, attendance)

ax.set_xlabel("Hours")
ax.set_ylabel("Marks")
ax.set_zlabel("Attendance")

plt.show()
```
