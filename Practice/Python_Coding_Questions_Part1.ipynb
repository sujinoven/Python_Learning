## 1.Write a program to take a name and age from the user and print a greeting message.

```python
def greet(name,age):

    print(f"Hello {name} and you are {age} years old")

namee = input("Enter Your Name")

agee = int(input("Enter Your Age:"))

greet(namee,agee)
```

## 2.Create a list of numbers and print only even numbers using a for loop.

```python
numbers = range(1,51)

even_numbers=[]

for i in numbers:
    if i % 2==0:
        even_numbers.append(i)
print(even_numbers)
```

```python
numbers = range(1, 51)

result = list(filter(lambda x: x % 2 == 0, numbers))

print(result)
```

### 3. Write a program that handles division by zero using try-except.

```python
a = int(input("Enter a first Number"))
b= int(input("Enter a Second Number:"))
try:
    result = a/b
    print(result)

except ZeroDivisionError:
    print("Can't Divide by Zero.Please give a Non Zero Number")
finally:
    print("The program ended!!")
```

### 4. Given a string, count how many vowels it has.

```python
string = "Hi Sujin!! How are You"

vowels=["a","e","i","o","u","A","E","I","O","U"]

count=0

for i in string:
    if i in vowels:

        count+=1

print(count)
```
