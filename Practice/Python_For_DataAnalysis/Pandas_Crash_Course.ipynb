## Data Cleaning

```python
order_details["item_price"]=order_details["item_price"].str.replace("$","").astype(float)
```

```python
order_details.describe()
```

## Top 5 selling items individual revenue it made

```python
order_details.head()
```

```python
order_details["total_revenue"] = order_details["quantity"] * order_details["item_price"]
order_details["total_revenue"]
```

```python
order_details.head()
```

```python
top5_selling_revenue = order_details.groupby("item_name")["total_revenue"].sum().sort_values(ascending=False).head()

top5_selling_revenue
```

## Total Revenue of the top5 items sold

```python
total_revenue_top5_items_sold = top5_selling_revenue.sum()

total_revenue_top5_items_sold
```
