## Number guessing using for loop

```python
secret_number = 50

for i in range(0,7): 
    user_input =int(input("Enter the Guessed Number:"))

    if secret_number == user_input:
        print("The guess is right")
        break
    else:
        print("The guess is wrong")
```

## Number guessing using while  loop

```python
secret_number = 50

while True:
    user_input =int(input("Enter the Guessed Number:"))

    if secret_number == user_input:
        print("The guess is right")
        break
    else:
        print("The guess is wrong")
```

## Number guessing with Nested If loop

```python
print("Welcome to the Number Guessing Game!!")

secret_number = 90

for i in range(1,8):

    user_input = int(input("Enter your guessed Number:"))

    if user_input < 1 or user_input >100:
        print("Invalid Number!! The number should be between 1 to 100")
    elif user_input < secret_number:
        print("Your number is less than Secret number")
    elif user_input > secret_number:
        print("Your Number is Greater than Secret Number")

    
    else:
        print("The Guessed  Number is correct")
        print(f"You found the {secret_number} in {i} attempts")
        break
```

## The Secret Number Should  be random everytime

```python
from random import randint

print("Welcome to the Number Guessing Game!!")

secret_number = randint(0,101)

for i in range(1,8):

    user_input = int(input("Enter your guessed Number:"))

    if user_input < 1 or user_input >100:
        print("Invalid Number!! The number should be between 1 to 100")
    elif user_input < secret_number:
        print("Your number is less than Secret number")
    elif user_input > secret_number:
        print("Your Number is Greater than Secret Number")

    

    elif user_input == secret_number:
        print("The Guessed  Number is correct")
        break
if user_input == secret_number:
   print(f"You found the {secret_number} in {i} attempts")

else:
        print("Unfortunately you didn't find the secret number in 7 attempts")
```

## Handling Errors

```python
from random import randint

print("Welcome to the Number Guessing Game!!")

secret_number = randint(0,101)

for i in range(1,8):
    try:
        user_input = int(input("Enter your guessed Number:"))

        if user_input < 1 or user_input >100:
            print("Invalid Number!! The number should be between 1 to 100")
        elif user_input < secret_number:
            print("Your number is less than Secret number")
        elif user_input > secret_number:
            print("Your Number is Greater than Secret Number")

        

        elif user_input == secret_number:
            print("The Guessed  Number is correct")
            break
    except ValueError:
        print("Please enter the values as Integer")
if user_input == secret_number:
   print(f"You found the {secret_number} in {i} attempts")

else:
        print("Unfortunately you didn't find the secret number in 7 attempts")
```

## Defining a function

```python
from random import randint
def guess_number():
    print("Welcome to the Number Guessing Game!!")

    secret_number = randint(0,101)

    for i in range(1,8):
        try:
            user_input = int(input("Enter your guessed Number:"))

            if user_input < 1 or user_input >100:
                print("Invalid Number!! The number should be between 1 to 100")
            elif user_input < secret_number:
                print("Your number is less than Secret number")
            elif user_input > secret_number:
                print("Your Number is Greater than Secret Number")

            

            elif user_input == secret_number:
                print("The Guessed  Number is correct")
                break
        except ValueError:
            print("Please enter the values as Integer")
    if user_input == secret_number:
        print(f"You found the {secret_number} in {i} attempts")

    else:
            print("Unfortunately you didn't find the secret number in 7 attempts")

            

```
